````python
# we have to use system prompt to provide a character or basically a personallity to our ai model 
#other wise it will not be specific to dsa.

#  we divide every row (data) 
# 1) System-> takes the prompt what we gave(be a dsa expert)
# 2) User->    Takes the question from the user
# 3) assistant-> we give it the solution that " ok give this as an output whenever asked"

# we have to divide each data row into 3 parts (above) so that the chat template or model can clearly distinguish bw the three

# use hf's Dataset to combine the datasets...
````

````python
SYSTEM_PROMPT="You are an expert DSA (Data Structures & Algorithms) tutor and competitive programmer. Explain your approach and provide clean code."
````

````python
unified_data=[]

print("Processing the Nvidia Dataset...")
for row in nvidia_data:
    # every iteration creates a separate dictionary for each row of data
    chat_format={
        # chat_format is a list of dictionaries(messages)
        "messages":[
            {"role":"system","content":SYSTEM_PROMPT},
            {"role":"user","content":row["input"]},# input as questions for nvidia is of the input format
            {"role":"assistant","content":row["output"]}# output as questions for nvidia is of the output format
        ]
    }
    unified_data.append(chat_format)
final_mega_dataset=Dataset.from_list(unified_data)
final_mega_dataset=final_mega_dataset.shuffle(seed=42)

print(f"After appending only Nvidia the length is :{len(final_mega_dataset)}\n")

print("Processing the codeforces Dataset...")
    
for row in code_forces_data:
    chat_format={
        "messages":[
            {"role":"system","content":SYSTEM_PROMPT},
            {"role":"user","content":row["prompt"]},
            {"role":"assistant","content":row["generation"]}

        ]
    }
    unified_data.append(chat_format)
final_mega_dataset=Dataset.from_list(unified_data)
final_mega_dataset=final_mega_dataset.shuffle(seed=42)

print(f"unified data is complete with a length  : {len(final_mega_dataset)}")
````

````python
from huggingface_hub import login
from datasets import load_dataset, Dataset, concatenate_datasets

# --- login (Kaggle secret) ---
from kaggle_secrets import UserSecretsClient
login(token=UserSecretsClient().get_secret("HF_TOKEN"))

HUGGING_FACE_USERNAME = "GAnirudh711"
SYSTEM_PROMPT = "You are an expert DSA (Data Structures & Algorithms) tutor and competitive programmer. Explain your approach and provide clean code."

def clean(t):
    return t.replace("\\n", "\n").replace("\\'", "'").replace('\\"', '"')

# --- 1. reload your V1 dataset (Nvidia + Codeforces) ---
v1 = load_dataset(f"{HUGGING_FACE_USERNAME}/DSA_FINETUNING_LLM", split="train")
is_cf = lambda r: r["messages"][1]["content"].startswith("You will be given a competitive programming problem")
cf = v1.filter(is_cf).shuffle(seed=42).select(range(500))
nv = v1.filter(lambda r: not is_cf(r)).shuffle(seed=42).select(range(1500))

# --- 2. reload LeetCode explanations ---
lc_stream = load_dataset("kaysss/leetcode-problem-solutions", split="train", streaming=True)
lc_raw = Dataset.from_list(list(lc_stream.take(10000)))
lc_raw = lc_raw.filter(lambda r: r["content"] and r["upvotes"] is not None
                       and "```" in r["content"] and len(r["content"]) < 6000)
lc_raw = lc_raw.sort("upvotes", reverse=True)
lc_raw = lc_raw.select(range(min(3000, len(lc_raw))))

lc_rows = [{"messages": [
    {"role": "system", "content": SYSTEM_PROMPT},
    {"role": "user", "content": f"Explain the approach for the LeetCode problem '{r['title']}', then give time and space complexity and code."},
    {"role": "assistant", "content": clean(r["content"])},
]} for r in lc_raw]
lc_ds = Dataset.from_list(lc_rows)

# --- 3. merge, shuffle, push ---
v2 = concatenate_datasets([nv, cf, lc_ds]).shuffle(seed=42)
print("V2 size:", len(v2), "| LeetCode rows:", len(lc_ds))
v2.push_to_hub(f"{HUGGING_FACE_USERNAME}/DSA_FINETUNING_LLM_V2")
````
